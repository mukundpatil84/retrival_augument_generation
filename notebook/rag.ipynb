```python
# Load the text document
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("../documents/resume.pdf")
documents = loader.load()

print(f"Actual Data:\n{documents[0].page_content}\n")
print(f"Meta Data:\n{documents[0].metadata}\n")
# print(documents)
```

```python
from langchain_text_splitters import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
        separator="\n",
        chunk_size=200,
        chunk_overlap=30,
        length_function=len,
        is_separator_regex=False,
    )
chunks = text_splitter.create_documents([documents[0].page_content])
print(chunks) 
# print(chunks[1]) 
print(len(chunks))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", "."],
    chunk_size=100, 
    chunk_overlap=20
    )
docs = text_splitter.split_documents(documents)
print(len(docs))
print(docs[:30])
```

```python
from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from uuid import uuid4

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vector_store = Chroma(
    collection_name="rag_collection",
    embedding_function=embeddings,
    persist_directory="../vector_store/chroma_langchain_db",  # Where to save data locally, remove if not necessary
)
uuids = [str(uuid4) for _ in range(len(docs)) ]
# vector_store.add_documents(docs, ids = uuids)
vector_store.add_documents(docs)
```

```python
question1 = "How much cost savings achieved through AWS Savings Plans and right-sizing compute?"
question2 = "Provide the details of the education completed?"
question3 = "Provide the details about the certification?"
similar_docs = vector_store.similarity_search(question1, k=1)
similar_docs[0].page_content
```

```python
similar_docs = vector_store.similarity_search(question2, k=1)
similar_docs[0].page_content
```

```python
# from langchain.chain import RetrievalQAWithSourceChain

from langchain_classic.chains import RetrievalQAWithSourcesChain 
from langchain_groq import ChatGroq
import dotenv

dotenv.load_dotenv()
 
groq_llm = ChatGroq(
    temperature=0.0,
    model_name="llama-3.3-70b-versatile"
)
qa_chain = RetrievalQAWithSourcesChain.from_llm(
    llm=groq_llm, 
    retriever=vector_store.as_retriever()
)
result = qa_chain.invoke({"question": question1},
    return_only_outputs=True
)

# result = qa_chain({"question": "How much cost savings achieved through AWS Savings Plans and right-sizing compute?"})
result
```

```python
from langchain_ollama import OllamaLLM

retriever=vector_store.as_retriever()

retrieved_docs = retriever.invoke(question3)
context = "\n\n".join([doc.page_content for doc in retrieved_docs])
prompt = f"Use the following context to answer the question:\n{context}\n\nQuestion: {question3}\nAnswer:"

ollam_model = OllamaLLM(
    model="llama3:latest",
    temperature=0.7,
    num_predict=1000   
)
response = ollam_model.invoke(prompt)
response
```
